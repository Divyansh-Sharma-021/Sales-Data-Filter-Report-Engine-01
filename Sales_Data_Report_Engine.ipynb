```python
import numpy as np
import pandas as pd

def display(value):
    if isinstance(value, pd.DataFrame):
        print(value.to_string())
    else:
        print(value)
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

np.random.seed(42)
sales = np.random.randint(2000, 20000, size=(10, 12))

sales_table = pd.DataFrame(
    sales,
    index=[f"Store {i}" for i in range(1, 11)],
    columns=months
)
```

```python
print("Sales Data")
print()
display(sales_table)
```

```python
quarter_1 = sales_table.iloc[:, 0:3]
print("Quarter 1")
display(quarter_1)
```

```python
quarter_2 = sales_table.iloc[:, 3:6]
print("Quarter 2")
display(quarter_2)
```

```python
quarter_3 = sales_table.iloc[:, 6:9]
print("Quarter 3")
display(quarter_3)
```

```python
quarter_4 = sales_table.iloc[:, 9:12]
print("Quarter 4")
display(quarter_4)
```

```python
print("Individual Sales — Store 1")
display(sales_table.iloc[[0]])
```

```python
target = 10000
print(f"Stores with monthly sales above {target}")

rows, cols = np.where(sales > target)

sales_over_target = pd.DataFrame({
    "Store": [f"Store {row + 1}" for row in rows],
    "Month": [months[col] for col in cols],
    "Sales": sales[rows, cols]
})

print(sales_over_target.to_string(index=False))
```
