```python
import numpy as np
import pandas as pd

def display(value):
    if isinstance(value, pd.DataFrame):
        print(value.to_string())
    else:
        print(value)
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = np.random.randint(2000, 20000, size=(10, 12))

sales_table = pd.DataFrame(
    sales,
    index=[f"Store {i}" for i in range(1, 11)],
    columns=months
)

print("Sales Data")
print()
display(sales_table)

Quater_1 = sales_table.iloc[:, 0:3]
Quater_2 = sales_table.iloc[:, 3:6]
Quater_3 = sales_table.iloc[:, 6:9]
Quater_4 = sales_table.iloc[:, 9:12]


print()
print("Quarter 1 ")
display(Quater_1)


print()
print("Quarter 2 ")
display(Quater_2)


print()
print("Quarter 3 ")
display(Quater_3)


print()
print("Quarter 4 ")
display(Quater_4)


print()
print("Individual Sales")
print()
display(sales_table.iloc[[0]])

target = 10000

print()
print("Target = 10000")
print("Stores who completed the target ")
rows, cols = np.where(sales > target)
sales_over_target = pd.DataFrame({
    "Store": [f"Store {r+1}" for r in rows],
    "Month": [months[c] for c in cols],
    "Sales": sales[rows, cols]})
print(sales_over_target.to_string(index=False))
```
